## DataFrames

A DataFrame is the main data structure in Pandas.

It is basically a table.

```python
# Approach 1: Create a DataFrame from a list of lists
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]

df = pd.DataFrame(data, columns=columns)
df
```

```python
# Approach 2: Create a DataFrame from a list of dictionaries
data = [
    {
        "Make": "Nissan",
        "Model": "Stanza",
        "Year": 1991,
        "Engine HP": 138.0,
        "Engine Cylinders": 4,
        "Transmission Type": "MANUAL",
        "Vehicle_Style": "sedan",
        "MSRP": 2000
    },
    {
        "Make": "Hyundai",
        "Model": "Sonata",
        "Year": 2017,
        "Engine HP": None,
        "Engine Cylinders": 4,
        "Transmission Type": "AUTOMATIC",
        "Vehicle_Style": "Sedan",
        "MSRP": 27150
    },
    # ...
]

df2 = pd.DataFrame(data)
df2
```

## Series

A series represents one column of a DataFrame

```python
# Accessing a series (column) from the DataFrame
# Approach 1: Dot notation
df.Make
```

```python
# Approach 2: Bracket notation
# More flexible, especially when column names have spaces or special characters
df['Make']
```

```python
# Access multiple columns
df[['Make', 'Model', 'MSRP']]
```

```python
# Add a new column
df['id'] = range(1, len(df) + 1)

df[['id', 'Make', 'Model', 'MSRP']]
```

```python
# Remove a column
del df['id']

df
```

## Index

An index is basically just the label we assign to each row in a DataFrame

```python
df.index
```

```python
# All series in the DF have an index
df['Make'].index
```

## Accessing elements

We use `.loc[]` to access a DF's elements based on its index labels

```python
# Access a row by its index label
df.loc[1]
```

```python
# Access multiple rows by their indices
df.loc[[1, 2]]
```

```python
# Change the values of the index
df.index = ['a', 'b', 'c', 'd', 'e']

df
```

```python
# Access the DF's rows by their new index values
df.loc[['b', 'c']]
```

We can also use `.iloc[]` to access elements, but this time we use the element's integer position

```python
# Access elements based on their positional index
df.iloc[[1, 2, 4]]
```

We can use `reset_index()` to reset the DF back to its default integer indexes.

*Note: This operation creates a new DF, it does not modify the original DF*

```python
# reset index (keep the old index as a column)
df.reset_index()
```

```python
# reset index (remove the custom index)
df.reset_index(drop=True)
```

```python
# Even if we reset the index, the original DataFrame remains unchanged
df
```

```python
# When resetting the index, remember to store the result in a variable
df = df.reset_index(drop=True)
df
```

## Filtering

```python
# Situation: we need to filter cars that were manufactured in 2015 or later
df['Year']
```

```python
# Step 1: Define a condition
# Retuns a boolean series for each row in the DF
df['Year'] >= 2015
```

```python
# Step 2: Use the boolean series to filter the DataFrame
# It will only return the rows where the condition is True
df[df['Year'] >= 2015]
```

```python
# Filter by multiple conditions
# Get all Nissan cars that were manufactured in 2015 or later
# We need to use parentheses to group the conditions and use the & operator (bitwise AND)
df[(df['Make'] == 'Nissan') & (df['Year'] >= 2015)]
```

## String operations

These operations are especially useful for cleaning text data

```python
# Common data issue: categories have inconsistent capitalization
# This can cause duplicated categories when grouping or aggregating data
df['Vehicle_Style']
```

```python
# Standardize values by converting them to lowercase
# Don't forget to assign the result back to the column
df['Vehicle_Style'] = df['Vehicle_Style'].str.lower()

df['Vehicle_Style']
```

```python
# Replace whitespace with underscores
df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_')

df['Vehicle_Style']
```

```python
# We can also chain multiple string methods together
df['Vehicle_Style'] = df['Vehicle_Style'].str.lower().str.replace(' ', '_')

df['Vehicle_Style']
```

## Missing values

In ML, we CANNOT have missing values (NaNs), so we need to handle them accordingly

```python
# Generates a boolean DataFrame indicating missing values
df.isnull()
```

```python
# Better approach: check missing values by column
df.isnull().sum()
```

## Grouping

```python
df.groupby('Transmission Type')['MSRP'].mean().round(2)
```

```python
# Group and aggregate by multiple columns
df.groupby(['Make', 'Transmission Type'])['MSRP'].mean().round(2)
```

```python
# Different aggregation functions can be applied to different columns
df.groupby('Make').agg({
    'MSRP': 'mean',
    'Engine HP': 'max',
    'Engine Cylinders': 'min'
}).round(2)
```

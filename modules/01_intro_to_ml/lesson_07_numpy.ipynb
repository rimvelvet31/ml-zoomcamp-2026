## Creating arrays

```python
np.zeros(5)
```

```python
np.ones(5)
```

```python
# Create a numpy array filled with a specific value
np.full(5, 12)
```

```python
# Create a numpy array from a list
my_list = [1, 2, 3, 4, 5]
print(my_list)
print(type(my_list))

print("")

np_array = np.array(my_list)
print(np_array)
print(type(np_array))
```

```python
# Accessing elements in a NumPy array
print(np_array[0])
print(np_array[2])
print(np_array[-1])
```

```python
# Changing values in a NumPy array
np_array[0] = 10
print(np_array)
```

```python
# Generate a range of numbers
print(np.arange(10)) # Generate numbers from 0 to 9
print(np.arange(5, 15)) # Generate numbers from 5 to 14
print(np.arange(0, 20, 2)) # Generate even numbers from 0 to 18
```

```python
np.linspace(0, 1, 5) # Generate 5 evenly spaced numbers between 0 and 1
```

## Multi-dimensional arrays

```python
# Generate a matrix of zeros with 5 rows and 2 columns
np.zeros((5, 2))
```

```python
# Create a numpy array from a list of lists (2D array)
list_of_lists = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

matrix = np.array(list_of_lists)
matrix
```

```python
# Accessing elements in a NumPy array
print(matrix[0, 0])  # Access the element in the first row and first column
print(matrix[1, 2])  # Access the element in the second row and third column
print(matrix[2, 1])  # Access the element in the third row and second column
print(matrix[2, 2])  # Access the element in the third row and third column
```

```python
# Changing elements in a 2D NumPy array
matrix[1, 2] = 99  # Change the element in the second row and third column to 99

print(matrix)
```

```python
# Access only rows
print(matrix[0])  # Access the first row
```

```python
# Change an entire row
matrix[0] = [10, 20, 30]  # Change the first row

print(matrix)
```

```python
# Acess only columns
print(matrix[:, 1])  # Access the second column
```

```python
# Replace an entire column
matrix[:, 1] = [100, 200, 300] # Change the second column

print(matrix)
```

## Randomly generated arrays

```python
# Generate a 5x2 array of random numbers between 0 and 1 (uses the uniform distribution)
np.random.rand(5, 2)
```

```python
# Set the random seed for reproducibility
np.random.seed(42)

# Generate a 5x2 array of random numbers between 0 and 1 (always the same due to the seed)
np.random.rand(5, 2) 
```

```python
# Generate random numbers from a normal distribution (mean=0, std=1)
np.random.randn(5, 2)
```

```python
# Generate random integers
np.random.randint(0, 10, size=(5, 2))
```

## Element-wise operations

```python
arr = np.arange(5)
arr
```

```python
# You can directly perform operations on NumPy arrays
arr + 1
```

```python
# Avoids the need for loops and makes the code more efficient
arr * 2
```

```python
# More complex operations
# Key is to remember that this formula applies to each element in the array
arr2 = (10 + (arr * 2)) ** 2
arr2
```

```python
# You can even perform operations between two arrays of the same shape
arr3 = arr + arr2
arr3
```

## Comparison operations

```python
a = np.arange(5)
print(a)

# Returns a boolean array (True if the element is even, False if it is odd)
a % 2 == 0
```

```python
# Boolean indexing: select only the even numbers from the array
a[a % 2 == 0]
```
